Upon first inspection, two issues present themselves:
* Unit ID numbers stored as float values
* 14 000 null Unit ID values

First, NaN unit IDs will be removed.
Then, unit IDs will be converted to strings, not integers, since they serve as identification numbers, not numbers to do integer operations with. 
The index of the DataFrame will also be set to gateway ID.

```python
gateways1 = gateways # TODO safety measure, change to real dataset later 
gateways1 = gateways.dropna()
gateways1['unitId'] = gateways1['unitId'].astype(int) # this is to drop the decimal place
gateways1['unitId'] = gateways1['unitId'].astype(str)
gateways1 = gateways1.set_index("gatewayId")
gateways1
```

```python
from datetime import datetime

def parse_last_seen(value: str) -> datetime:
    """Parse one lastSeen value from any format used in gateways.csv."""
    formats = ("%Y-%m-%d", "%m/%d/%Y, %H:%M:%S", "%d %b %Y", "%Y/%m/%d")
    for date_format in formats:
        try:
            return datetime.strptime(value, date_format)
        except ValueError:
            continue
    raise ValueError(f"Unsupported lastSeen date: {value!r}")

gateways2 = gateways1.copy()
gateways2["lastSeen"] = pd.to_datetime(gateways2["lastSeen"].map(parse_last_seen))
gateways2[["lastSeen"]].head()
```
